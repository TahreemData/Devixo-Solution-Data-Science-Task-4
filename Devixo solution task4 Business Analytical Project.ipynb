```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
file_path = r"C:\Users\Diyam\Desktop\Jupiter Word\Global Superstore.csv"

df = pd.read_csv(file_path, encoding='latin1')

df.head()
```

```python
# missing value
df.isnull().sum()
```

```python
# Duplicated 
df.duplicated().sum()
```

```python
# Balancing  Missing Value in Poatal code
df['Postal Code']=df['Postal Code'].fillna('unknown')
df.isnull().sum()
```

```python
#No missing Value or no Duplicate value i this data
```

```python
# Checking dataset shape
print(df.shape)
# dataset information
df.info()
# statistical summary
df.describe()
# Columns Name
print(df.columns)
# First five rows
df.head()
```

```python
# Checking and Removing duplictes
print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
# Converting date columns
df['Order Date']=pd.to_datetime(df['Order Date'], errors='coerce')
df['Ship Date']=pd.to_datetime(df['Ship Date'], errors='coerce')
print(df[['Order Date', 'Ship Date']].dtypes)
```

```python
print("Missing Order Dates:", df['Order Date'].isnull().sum())
print("Missing Ship Dates:", df['Ship Date'].isnull().sum())
```

```python
# Check numerical columns
numeric_columns = [
    'Sales', 'Quantity', 'Discount',
    'Profit', 'Shipping Cost'
]

print(df[numeric_columns].describe())
```

```python
# Making valid values
print("Negative Sales:", (df['Sales'] < 0).sum())
print("Zero or Negative Quantity:", (df['Quantity'] <= 0).sum())
print("Negative Shipping Cost:", (df['Shipping Cost'] < 0).sum())
print("Discount outside 0-1:", 
      ((df['Discount'] < 0) | (df['Discount'] > 1)).sum())
```

```python
# feature engineering
# Shipping days
df['Shipping Days'] = (
    df['Ship Date'] - df['Order Date']
).dt.days

# Order year
df['Order Year'] = df['Order Date'].dt.year

# Order month
df['Order Month'] = df['Order Date'].dt.month

# Month name
df['Month Name'] = df['Order Date'].dt.month_name()

# Year and month
df['Year-Month'] = df['Order Date'].dt.to_period('M').astype(str)

# Profit margin
df['Profit Margin'] = np.where(
    df['Sales'] != 0,
    (df['Profit'] / df['Sales']) * 100,
    np.nan
)

df.head()
```

```python
# Final dataset shape
print("Final Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

df.info()
```

```python
output_path = r"C:\Users\Diyam\Desktop\Jupiter Word\Global_Superstore_Cleaned.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
```

```python
# Basic bussiness statistics
# Basic Business Statistics

print("Total Sales:", df['Sales'].sum())
print("Total Profit:", df['Profit'].sum())
print("Total Quantity:", df['Quantity'].sum())

print("Total Orders:", df['Order ID'].nunique())
print("Total Customers:", df['Customer ID'].nunique())
print("Total Products:", df['Product ID'].nunique())

print("Average Order Value:",
      df.groupby('Order ID')['Sales'].sum().mean())

print("Overall Profit Margin (%):",
      (df['Profit'].sum() / df['Sales'].sum()) * 100)
```

```python
# out;ier detection
outlier_columns = [
    'Sales',
    'Profit',
    'Quantity',
    'Discount',
    'Shipping Cost',
    'Shipping Days'
]

outlier_summary = []

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    outlier_summary.append({
        'Column': col,
        'Lower Limit': lower_limit,
        'Upper Limit': upper_limit,
        'Outlier Count': len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```
